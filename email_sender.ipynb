```python
#email sender
from email.message import EmailMessage
import smtplib

#For google go to account settings, search app passwords to get pass
APP_PASSWORD = "xxxx xxxx xxxx xxxx"

message = EmailMessage()
message["Subject"] = "Hello from Python PT"
message["To"] = "xxx@gmail.com"
message["From"] = "yyy@gmail.com"

#this is the message body
message.set_content("This is a test for my code!")

print(message)

#securely login with smtp and ssl
with smtplib.SMTP_SSL("smtp.gmail.com", 465) as email_server:
  email_server.login(
      "yyy@gmail.com",
      APP_PASSWORD
  )

  email_server.send_message(message)

print("email was sent ")









```
